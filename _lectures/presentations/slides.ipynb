Good presentations usually *do not* closely follow a paper

But a good presentation can often be expanded into a paper


# Structure


## Classic structure


* Introduction 
	 * Welcome and introduction
   * Say who you are, why you are there
   * Mention anything else that’s significant, i.e., when you will be taking questions
* Say what you’re going to say
   * Set the scene, provide background
   * Give the outline
* Say it
   * Get into more detail, give outline and analysis of the topic
   * Investigation
   * Issues
* Say what you said
   * Provide conclusions
   * Repeat key messages and provide a resolution

::: {.center}
*This is reliable and safe but may not engage audience*
:::


## Anything else



::: {.center}
*This is usually more engaging but requires more thought*
:::


:::
::::



* Introduction
   * Minimal logistics, e.g., who you are
   * Hook immediately

* Time is precious
   * Trim all fat
   * Q&A exists for a reason

* Don't fight your slides
   * Minimize words
   * Interact with images


# Style

* Humor
   * If you're funny, can keep people engaged
   * But risk not being taken seriously and hard to calibrate for many audiences and purposes
* Engagement
   * Can keep people more interested
   * Hard to predict and control
* Language
   * Even simpler/clearer than paper
   * Use jargon only when everyone will understand or you can define it
* Script or no script


# Visuals

* Don't display anything you don't want them to look at
* Visuals will compete for attention
* Either extremely simple or you walk people through it
